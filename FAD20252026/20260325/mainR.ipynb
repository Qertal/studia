# Funkcje tworzące bazy w R

Funkcje w R, które dla każdego systemu bazowego definiują zestaw ( K ) funkcji bazowych ( f_k ), nazywane są **funkcjami tworzącymi** (*create functions*).

## Dostępne funkcje tworzące

```r
basisobj = create.constant.basis(rangeval)
basisobj = create.monomial.basis(rangeval, nbasis)
basisobj = create.fourier.basis(rangeval, nbasis, period)
basisobj = create.bspline.basis(rangeval, nbasis, norder, breaks)
```

Skonstruowany system bazowy (`basisobj`) jest obiektem klasy `basisfd`.

---

## Najczęściej używane argumenty

* **rangeval** – zakres argumentu ( t ), wektor długości 2

  ```r
  rangeval = c(0,1)
  ```

* **nbasis** – liczba funkcji bazowych ( K )

(Uwaga: w bazie stałej ( K = 1 ) automatycznie)

---

# Baza Fouriera

### Przykład

```r
daybasis65 = create.fourier.basis(c(0,365), 65)
```

Domyślnie:
[
T = 365
]

Można jawnie podać okres:

```r
create.fourier.basis(c(0,365), 65, T)
```

---

## Uwagi

* Jeśli ( K ) jest parzyste → R automatycznie dodaje funkcję → ( K = K+1 )
* Zabezpieczenia przed osobliwością przy nadmiarze funkcji

---

### Szereg kosinusowy

Funkcje mają zerowe pochodne na końcach.

Można usunąć wybrane funkcje:

```r
dropind = c(...)
```

---

### Usunięcie wyrazu stałego (średnia = 0)

```r
zerobasis = create.fourier.basis(rangeval, nbasis, dropind=1)
```

lub:

```r
zerobasis = daybasis65[2:65]
```

---

## Pełna składnia

```r
create.fourier.basis(
    rangeval = c(0, 1),
    nbasis = 3,
    period = diff(rangeval),
    dropind = NULL,
    quadvals = NULL,
    values = NULL,
    basisvalues = NULL,
    names = NULL,
    axes = NULL
)
```

Pomoc:

```r
?create.fourier.basis
```

# Zadanie 1 – B-splajny

## Tworzenie bazy

```r
splinebasis = create.bspline.basis(c(0,10), 13)
plot(splinebasis)
```

---

## Własności

* Funkcje środkowe:

  * dodatnie na **4 podprzedziałach**
* Funkcje brzegowe:

  * dodatnie na mniejszym zakresie
* Każda funkcja:

  * rośnie → maksimum → maleje → 0

---

## Pojedyncza funkcja

```r
plot(splinebasis[6])
```

Obserwacje:

* rząd 4 → wsparcie na 4 przedziałach
* rząd 6 → wsparcie na 6 przedziałach
