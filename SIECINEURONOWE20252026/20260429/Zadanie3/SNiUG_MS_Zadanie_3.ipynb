# Modele Dyfuzyjne (DDPM) - Zadanie
### Wprowadzenie
W przeciwieństwie do sieci GAN, w których Generator i Dyskryminator rywalizują ze sobą w czasie rzeczywistym, modele dyfuzyjne opierają się na powolnym, iteracyjnym procesie inspirowanym termodynamiką. Podstawowa idea polega na zniszczeniu struktury danych poprzez stopniowe dodawanie szumu, a następnie wytrenowaniu sieci neuronowej, która nauczy się ten proces odwracać.

Architektura ta dzieli się na dwa główne etapy:

- Proces w przód (Forward / Noising Process): Ustalony z góry (nieuczący się) łańcuch Markowa, w którym do obrazu $x_0$ w każdym kroku $t$ dodawana jest niewielka ilość szumu Gaussowskiego. Po $T$ krokach (np. $T=1000$), obraz staje się całkowicie nierozróżnialnym, czystym szumem $x_T$.
- Proces w tył (Backward / Denoising Process): Sieć neuronowa (najczęściej oparta na architekturze U-Net) uczy się, jak usunąć szum dodany w jednym, konkretnym kroku czasowym $t$.

### Sztuczka z reparametryzacją 
Przejście przez cały łańcuch Markowa krok po kroku podczas treningu byłoby skrajnie niewydajne. Ponieważ suma wielu rozkładów normalnych to nadal rozkład normalny, możemy wyliczyć matematycznie parametry szumu, aby w jednym kroku przeskoczyć z czystego obrazka $x_0$ od razu do dowolnego zaszumionego stanu $x_t$. Do sterowania tempem dodawania szumu wykorzystujemy harmonogram wariancji (variance schedule) oznaczany jako $\beta_t$.
