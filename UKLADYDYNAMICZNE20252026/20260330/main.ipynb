## Zadanie 1

Rozważamy układ dynamiczny
$$
x_{n+1}=f(x_n), \qquad f:[0,\infty)\to[0,\infty), \qquad f(x)=\sqrt[3]{x}.
$$

Mamy:

* wyznaczyć punkty stałe,
* zbadać ich stabilność,
* opisać zbieżność trajektorii dla dowolnego $x_0\in[0,\infty)$,
* wyznaczyć baseny przyciągania,
* a następnie rozważyć ograniczenie do przedziału $\left[\frac12,\infty\right)$.

---

## a) Punkty stałe

Szukamy rozwiązań równania
$$
f(x)=x.
$$

Czyli
$$
\sqrt[3]{x}=x.
$$

Podnosząc obie strony do trzeciej potęgi, dostajemy
$$
x=x^3,
$$
czyli
$$
x^3-x=0,
$$
stąd
$$
x(x-1)(x+1)=0.
$$

Ponieważ rozważamy dziedzinę $[0,\infty)$, zostają tylko:
$$
Fix(f)={0,1}.
$$

---

## b) Stabilność punktów stałych

Dla $x>0$ mamy
$$
f'(x)=\left(x^{1/3}\right)'=\frac13 x^{-2/3}.
$$

### Punkt stały $x=1$

Liczymy
$$
f'(1)=\frac13.
$$

Zatem
$$
|f'(1)|=\frac13<1.
$$

Wniosek:
$$
1 \text{ jest atraktorem.}
$$

### Punkt stały $x=0$

W punkcie $0$ pochodna nie istnieje, więc nie można tu bezpośrednio zastosować prostego kryterium z pochodnej. Trzeba zbadać zachowanie trajektorii.

Pokażemy później, że dla każdego $x_0\in(0,1)$ mamy
$$
x_n\to 1,
$$
a nie do $0$. To oznacza, że punkt $0$ nie przyciąga punktów ze swojego otoczenia, lecz je „odpycha”.

Wniosek:
$$
0 \text{ jest repelerem.}
$$

---

## c) Zbieżność trajektorii dla dowolnego $x_0\in[0,\infty)$

Rozpatrzymy osobno kilka przypadków.

### Przypadek 1: $x_0=0$

Wtedy
$$
x_1=f(0)=0,
$$
więc
$$
x_n=0 \quad \text{dla każdego } n.
$$

Zatem
$$
x_n\to 0.
$$

---

### Przypadek 2: $x_0\in(0,1)$

Pokażemy najpierw, że dla $x\in(0,1)$ zachodzi
$$
f(x)>x.
$$

Rzeczywiście, rozważmy funkcję
$$
g(x)=f(x)-x=\sqrt[3]{x}-x.
$$

Mamy
$$
g(0)=0,\qquad g(1)=0.
$$

Ponadto
$$
g''(x)=\left(\frac13 x^{-2/3}-1\right)'=-\frac29 x^{-5/3}<0
\quad \text{dla } x\in(0,1).
$$

Zatem $g$ jest wklęsła na $(0,1)$, a skoro przyjmuje wartości $0$ na końcach przedziału, to dla punktów wewnętrznych jest dodatnia. Stąd
$$
g(x)>0 \quad \text{dla } x\in(0,1),
$$
czyli
$$
\sqrt[3]{x}>x \quad \text{dla } x\in(0,1).
$$

Wobec tego dla trajektorii:
$$
x_{n+1}=f(x_n)>x_n,
$$
więc ciąg $(x_n)$ jest rosnący.

Jednocześnie, ponieważ $x_0<1$, to
$$
x_1=\sqrt[3]{x_0}<1,
$$
a indukcyjnie
$$
x_n<1 \quad \text{dla każdego } n.
$$

Zatem $(x_n)$ jest:

* rosnący,
* ograniczony z góry przez $1$.

Wynika stąd, że ciąg jest zbieżny. Niech
$$
x_n\to g.
$$

Przechodząc do granicy w równaniu
$$
x_{n+1}=\sqrt[3]{x_n},
$$
dostajemy
$$
g=\sqrt[3]{g},
$$
czyli
$$
g\in{0,1}.
$$

Ponieważ cały ciąg leży w $(0,1)$ i jest rosnący, granica nie może być równa $0$. Zatem
$$
g=1.
$$

Czyli:
$$
x_0\in(0,1)\implies x_n\to 1.
$$

---

### Przypadek 3: $x_0=1$

Wtedy
$$
x_1=f(1)=1,
$$
więc
$$
x_n=1 \quad \text{dla każdego } n.
$$

Zatem
$$
x_n\to 1.
$$

---

### Przypadek 4: $x_0\in(1,\infty)$

Teraz pokażemy, że dla $x>1$ zachodzi
$$
f(x)<x.
$$

Ponownie bierzemy
$$
g(x)=\sqrt[3]{x}-x.
$$

Mamy
$$
g(1)=0,
$$
a także
$$
g'(x)=\frac13 x^{-2/3}-1.
$$

Dla $x>1$ mamy $x^{-2/3}<1$, więc
$$
\frac13 x^{-2/3}<\frac13,
$$
czyli
$$
g'(x)<0.
$$

Zatem $g$ jest malejąca na $(1,\infty)$, a ponieważ $g(1)=0$, to dla $x>1$ mamy
$$
g(x)<0.
$$

Stąd
$$
\sqrt[3]{x}-x<0,
$$
czyli
$$
\sqrt[3]{x}<x \quad \text{dla } x>1.
$$

Wtedy
$$
x_{n+1}=f(x_n)<x_n,
$$
więc ciąg $(x_n)$ jest malejący.

Ponadto, jeśli $x_n>1$, to
$$
x_{n+1}=\sqrt[3]{x_n}>1.
$$

Zatem $(x_n)$ jest:

* malejący,
* ograniczony z dołu przez $1$.

A więc jest zbieżny. Niech
$$
x_n\to g.
$$

Przechodząc do granicy w równaniu
$$
x_{n+1}=\sqrt[3]{x_n},
$$
otrzymujemy
$$
g=\sqrt[3]{g},
$$
czyli znów
$$
g\in{0,1}.
$$

Ponieważ $x_n\ge 1$, granica spełnia $g\ge 1$, więc
$$
g=1.
$$

Czyli:
$$
x_0\in(1,\infty)\implies x_n\to 1.
$$

---

## Podsumowanie punktu c)

Dla każdego $x_0\in[0,\infty)$ mamy:

$$
\begin{cases}
x_0=0 &\Longrightarrow x_n\to 0,\
x_0\in(0,1) &\Longrightarrow x_n\to 1,\
x_0=1 &\Longrightarrow x_n\to 1,\
x_0\in(1,\infty) &\Longrightarrow x_n\to 1.
\end{cases}
$$

Czyli jedynie trajektoria startująca z $0$ pozostaje w $0$, a wszystkie pozostałe trajektorie dążą do $1$. Taki opis jest zgodny z rozwiązaniem z materiałów. 

---

## d) Baseny przyciągania punktów stałych

### Basen przyciągania punktu $1$

Do punktu $1$ dążą wszystkie trajektorie poza tą startującą z $0$. Zatem
$$
Bas_f(1)=(0,1)\cup{1}\cup(1,\infty)=(0,\infty).
$$

Jeżeli w Twoich notatkach basen definiowany jest bez samego punktu orbity granicznej, to można też zapisać
$$
Bas_f(1)=(0,1)\cup(1,\infty).
$$

To jest kwestia przyjętej konwencji. W praktyce z dynamiki wynika, że również punkt $1$ należy do własnego basenu przyciągania, jeśli dopuszczamy punkty z orbity stałej.

### Basen przyciągania punktu $0$

Jedynie dla $x_0=0$ dostajemy zbieżność do $0$, więc
$$
Bas_f(0)={0}.
$$

---

## Punkty okresowe

Z otrzymanego opisu dynamiki wynika, że nie ma innych orbit okresowych niż punkty stałe $0$ i $1$.

Rzeczywiście, każda trajektoria:

* albo od razu stoi w punkcie $0$,
* albo dąży do $1$.

Nie ma więc miejsca na niebanalną orbitę okresową. Zgodnie z lematem o pełności opisu dynamiki z materiałów, zbiór orbit okresowych jest zatem wyczerpany przez punkty stałe. 

---

## e) Przypadek

$$
f:\left[\frac12,\infty\right)\to\left[\frac12,\infty\right), \qquad f(x)=\sqrt[3]{x}
$$

Teraz dziedzina została zawężona do
$$
\left[\frac12,\infty\right).
$$

### Punkty stałe

Równanie punktów stałych pozostaje takie samo:
$$
\sqrt[3]{x}=x.
$$

W poprzednim przypadku dostaliśmy punkty $0$ i $1$, ale teraz $0\notin\left[\frac12,\infty\right)$, więc zostaje tylko
$$
Fix(f)={1}.
$$

---

### Stabilność

Dla $x=1$ nadal mamy
$$
f'(1)=\frac13,
$$
czyli
$$
|f'(1)|<1.
$$

Zatem
$$
1 \text{ jest atraktorem.}
$$

---

### Zbieżność trajektorii

Weźmy dowolne
$$
x_0\in\left[\frac12,\infty\right).
$$

Są dwa przypadki:

* jeśli $x_0\in\left[\frac12,1\right)$, to tak jak wcześniej mamy $x_n$ rosnący i ograniczony z góry przez $1$, więc
  $$
  x_n\to 1;
  $$

* jeśli $x_0>1$, to tak jak wcześniej mamy $x_n$ malejący i ograniczony z dołu przez $1$, więc
  $$
  x_n\to 1;
  $$

* jeśli $x_0=1$, to oczywiście
  $$
  x_n=1.
  $$

Zatem dla każdego
$$
x_0\in\left[\frac12,\infty\right)
$$
zachodzi
$$
x_n\to 1.
$$

---

### Basen przyciągania

W tej nowej przestrzeni cały zbiór
$$
\left[\frac12,\infty\right)
$$
jest basenem przyciągania punktu $1$:
$$
Bas_f(1)=\left[\frac12,\infty\right).
$$

Nie ma innych punktów stałych ani innych orbit okresowych.

---

## Ostateczny wniosek

Dla układu
$$
x_{n+1}=\sqrt[3]{x_n}, \qquad x_n\in[0,\infty),
$$
mamy:

* punkty stałe:
  $$
  Fix(f)={0,1};
  $$

* stabilność:

  * $0$ jest repelerem,
  * $1$ jest atraktorem;

* zbieżność trajektorii:
  $$
  x_0=0 \implies x_n\to 0,
  $$
  $$
  x_0>0 \implies x_n\to 1;
  $$

* baseny przyciągania:
  $$
  Bas_f(0)={0}, \qquad Bas_f(1)=(0,\infty)
  $$
  lub, zależnie od konwencji,
  $$
  Bas_f(1)=(0,1)\cup(1,\infty);
  $$

* innych orbit okresowych nie ma.

Dla ograniczenia do
$$
\left[\frac12,\infty\right)
$$
jedynym punktem stałym jest $1$, jest on atraktorem, a każda trajektoria dąży do $1$.


## Zadanie 2

Rozważamy układ dynamiczny
$$
x_{n+1}=f(x_n), \qquad f:\mathbb{C}\to\mathbb{C}, \qquad f(x)=\frac{x+1}{5}.
$$

Mamy:

* wyznaczyć punkt stały,
* zbadać jego stabilność,
* udowodnić zbieżność każdej trajektorii,
* wyznaczyć basen przyciągania,
* uzasadnić zbieżność przez jawną postać ciągu.

---

## a) Zbiór punktów stałych $Fix(f)$

Szukamy rozwiązań równania
$$
f(x)=x.
$$

Czyli
$$
\frac{x+1}{5}=x.
$$

Mnożąc stronami przez $5$, dostajemy
$$
x+1=5x.
$$

Stąd
$$
1=4x,
$$
więc
$$
x=\frac14.
$$

Zatem
$$
Fix(f)=\{\frac14\}.
$$

---

## b) Stabilność punktu stałego

Liczymy pochodną:
$$
f'(x)=\frac15.
$$

W szczególności dla punktu stałego $x_0=\frac14$ mamy
$$
\left|f'\left(\frac14\right)\right|=\frac15<1.
$$

Z kryterium stabilności wynika więc, że punkt stały
$$
\frac14
$$
jest **atraktorem**. 

---

## c) Dowód, że dla każdego $x_0\in\mathbb{C}$ trajektoria $(x_n)_{n\geq 0}$ zbiega do punktu stałego

Chcemy pokazać, że dla dowolnego $x_0\in\mathbb{C}$ ciąg zadany rekurencyjnie przez
$$
x_{n+1}=\frac{x_n+1}{5}
$$
spełnia
$$
x_n\to \frac14.
$$

Najwygodniej odjąć od obu stron punkt stały. Niech
$$
y_n=x_n-\frac14.
$$

Wtedy
$$
y_{n+1}=x_{n+1}-\frac14
=\frac{x_n+1}{5}-\frac14.
$$

Sprowadzamy do wspólnego mianownika:
$$
y_{n+1}
=\frac{4x_n+4-5}{20}
=\frac{4x_n-1}{20}.
$$

Ponieważ
$$
4x_n-1=4\left(x_n-\frac14\right)=4y_n,
$$
to dostajemy
$$
y_{n+1}=\frac{4y_n}{20}=\frac15 y_n.
$$

A więc
$$
y_n=\left(\frac15\right)^n y_0.
$$

Ponieważ
$$
\left|\frac15\right|<1,
$$
mamy
$$
\left(\frac15\right)^n\to 0,
$$
czyli
$$
y_n\to 0.
$$

Zatem
$$
x_n-\frac14\to 0,
$$
a więc
$$
x_n\to \frac14.
$$

To kończy dowód.

---

## d) Wyznaczenie zbioru $Bas_f$

Ponieważ dla **każdego** $x_0\in\mathbb{C}$ trajektoria zbiega do punktu stałego $\frac14$, basen przyciągania tego punktu jest całym zbiorem $\mathbb{C}$.

Zatem
$$
Bas_f\left(\frac14\right)=\mathbb{C}.
$$

Jeżeli przez $Bas_f$ rozumiemy zbiór wszystkich punktów, których trajektorie są przyciągane przez jakiś atraktor, to tutaj po prostu
$$
Bas_f=\mathbb{C}.
$$

---

## e) Jawna postać ciągu $(x_n)$ i uzasadnienie zbieżności

Z poprzedniego punktu mieliśmy
$$
x_n-\frac14=\left(\frac15\right)^n\left(x_0-\frac14\right).
$$

Stąd otrzymujemy jawną postać:
$$
x_n=\frac14+\left(\frac15\right)^n\left(x_0-\frac14\right).
$$

Można też to rozpisać jako
$$
x_n=\left(\frac15\right)^n x_0+\frac14\left(1-\left(\frac15\right)^n\right).
$$

Teraz zbieżność jest już bezpośrednio widoczna, bo
$$
\left(\frac15\right)^n\to 0.
$$

Wobec tego
$$
x_n\to \frac14+0\cdot\left(x_0-\frac14\right)=\frac14.
$$

Czyli dla każdego $x_0\in\mathbb{C}$:
$$
x_n\to \frac14.
$$

---

## Ostateczny wniosek

Dla układu
$$
x_{n+1}=\frac{x_n+1}{5}
$$
zachodzi:

$$
Fix(f)=\{1/4\},
$$

punkt stały $\frac14$ jest atraktorem, ponieważ
$$
|f'(x)|=\frac15<1,
$$
a dla każdego $x_0\in\mathbb{C}$ trajektoria zbiega do tego punktu.

Jawna postać ciągu to
$$
x_n=\frac14+\left(\frac15\right)^n\left(x_0-\frac14\right),
$$
więc
$$
x_n\to \frac14.
$$

Basen przyciągania wynosi
$$
Bas_f\left(\frac14\right)=\mathbb{C}.
$$


## Zadanie 3

Rozważamy układ dynamiczny
$$
x_{n+1}=f(x_n), \qquad f:\mathbb{C}\to\mathbb{C}, \qquad f(x)=22x+7.
$$

Mamy:

* wyznaczyć punkt stały,
* zbadać jego stabilność,
* pokazać, że poza punktem stałym trajektorie uciekają do nieskończoności,
* zidentyfikować atraktor,
* wyprowadzić jawną postać ciągu.

---

## a) Zbiór punktów stałych $Fix(f)$

Szukamy rozwiązań równania
$$
f(x)=x.
$$

Czyli
$$
22x+7=x.
$$

Stąd
$$
21x=-7,
$$
więc
$$
x=-\frac{1}{3}.
$$

Zatem
$$
Fix(f)=\{-\frac13\}.
$$

---

## b) Stabilność punktu stałego

Liczymy pochodną:
$$
f'(x)=22.
$$

W szczególności dla punktu stałego $x_0=-\frac13$ mamy
$$
\left|f'\left(-\frac13\right)\right|=22>1.
$$

Zatem punkt stały
$$
-\frac13
$$
jest **repelerem**. 

---

## c) Dowód, że dla każdego $x_0\in\mathbb{C}\setminus\{-\frac13\}$ trajektoria $(x_n)_{n\geq 0}$ ucieka do nieskończoności

Najwygodniej znów odjąć punkt stały. Niech
$$
y_n=x_n+\frac13.
$$

Wtedy
$$
y_{n+1}=x_{n+1}+\frac13=22x_n+7+\frac13.
$$

Ponieważ
$$
7+\frac13=\frac{22}{3},
$$
mamy
$$
y_{n+1}=22x_n+\frac{22}{3}=22\left(x_n+\frac13\right)=22y_n.
$$

Zatem
$$
y_n=22^n y_0 = 22^n\left(x_0+\frac13\right).
$$

Czyli
$$
x_n+\frac13 = 22^n\left(x_0+\frac13\right),
$$
a więc
$$
x_n = -\frac13 + 22^n\left(x_0+\frac13\right).
$$

Jeśli
$$
x_0\neq -\frac13,
$$
to
$$
x_0+\frac13\neq 0.
$$

Wtedy
$$
|x_n+\tfrac13| = 22^n,\left|x_0+\tfrac13\right|.
$$

Ponieważ $22^n\to\infty$ oraz $\left|x_0+\tfrac13\right|>0$, dostajemy
$$
|x_n+\tfrac13|\to\infty.
$$

Stąd również
$$
|x_n|\to\infty.
$$

Zatem dla każdego
$$
x_0\in\mathbb{C}\setminus\{-\frac13\}
$$
trajektoria ucieka do nieskończoności.

---

## d) Identyfikacja atraktora układu

Tu trzeba uważać: **układ nie ma atraktora w $\mathbb{C}$**.

Jedyny punkt stały to
$$
-\frac13,
$$
ale jest on repelerem, nie atraktorem.

Dodatkowo każda trajektoria startująca z punktu różnego od $-\frac13$ ucieka do nieskończoności, więc w przestrzeni $\mathbb{C}$ nie ma zbioru, który przyciągałby otoczenie.

Zatem:

$$
\text{układ nie posiada atraktora w } \mathbb{C}.
$$

Jeżeli prowadzący dopuszcza odpowiedź w rozszerzonej przestrzeni zespolonej $\mathbb{C}\cup{\infty}$, to można powiedzieć, że rolę globalnego atraktora pełni $\infty$. Ale w zwykłej przestrzeni $\mathbb{C}$ taka odpowiedź byłaby zbyt swobodna, więc lepiej napisać wprost, że **brak atraktora w $\mathbb{C}$**.

---

## e) Jawna postać ciągu $(x_n)$ i uzasadnienie jego rozbieżności

Z punktu c) otrzymaliśmy już wzór jawny:
$$
x_n=-\frac13+22^n\left(x_0+\frac13\right).
$$

To jest jawna postać ciągu.

Teraz:

* jeśli
  $$
  x_0=-\frac13,
  $$
  to
  $$
  x_n=-\frac13 \quad \text{dla każdego } n,
  $$
  więc ciąg jest stały;

* jeśli
  $$
  x_0\neq -\frac13,
  $$
  to
  $$
  x_0+\frac13\neq 0,
  $$
  więc składnik
  $$
  22^n\left(x_0+\frac13\right)
  $$
  ma moduł dążący do nieskończoności.

W konsekwencji
$$
|x_n|\to\infty,
$$
czyli ciąg jest rozbieżny w $\mathbb{C}$ i ucieka do nieskończoności.

---

## Ostateczny wniosek

Dla układu
$$
x_{n+1}=22x_n+7
$$
zachodzi:

$$
Fix(f)=\{-\frac13\},
$$

a jedyny punkt stały jest repelerem, ponieważ
$$
|f'(x)|=22>1.
$$

Jawna postać ciągu jest dana wzorem
$$
x_n=-\frac13+22^n\left(x_0+\frac13\right).
$$

Zatem:

* dla
  $$
  x_0=-\frac13
  $$
  mamy trajektorię stałą,

* dla
  $$
  x_0\neq -\frac13
  $$
  mamy
  $$
  |x_n|\to\infty.
  $$

W szczególności układ **nie ma atraktora w $\mathbb{C}$**.

## Zadanie 4

Rozważamy funkcję
$$
f:[0,1]\to[0,1], \qquad
f(x)=
\begin{cases}
\dfrac{x}{\pi}, & x\in\mathbb{Q},[1ex]
x, & x\in\mathbb{R}\setminus\mathbb{Q}.
\end{cases}
$$

Mamy:

* wyznaczyć zbiór punktów stałych,
* zbadać, które z nich są atraktorami lub repelerami,
* dla każdego $x\in[0,1]$ określić zbiór $Bas_f(x)$.

---

## a) Wyznaczenie zbioru punktów stałych $Fix(f)$

Szukamy wszystkich $x\in[0,1]$, dla których
$$
f(x)=x.
$$

Rozpatrzmy dwa przypadki.

### Przypadek 1: $x\in\mathbb{Q}$

Wtedy
$$
f(x)=\frac{x}{\pi}.
$$

Warunek punktu stałego ma postać
$$
\frac{x}{\pi}=x.
$$

Czyli
$$
x\left(\frac{1}{\pi}-1\right)=0.
$$

Ponieważ $\frac{1}{\pi}-1\neq 0$, otrzymujemy
$$
x=0.
$$

Zatem jedynym wymiernym punktem stałym jest $0$.

### Przypadek 2: $x\in\mathbb{R}\setminus\mathbb{Q}$

Wtedy
$$
f(x)=x,
$$
więc każdy punkt niewymierny z przedziału $[0,1]$ jest punktem stałym.

---

### Wniosek

Zbiór punktów stałych to
$$
Fix(f)=\left([0,1]\setminus\mathbb{Q}\right)\cup{0}.
$$

Czyli:

* $0$ jest punktem stałym,
* każdy punkt niewymierny z $[0,1]$ jest punktem stałym,
* każda dodatnia liczba wymierna z $[0,1]$ nie jest punktem stałym.

---

## b) Które punkty stałe są atraktorami lub repelerami?

Tu trzeba uważać, bo funkcja nie jest ciągła, więc nie można mechanicznie używać kryterium z pochodnej.

Przypomnijmy sens:

* punkt stały $x_0$ jest **atraktorem**, gdy istnieje jego otoczenie, z którego wszystkie trajektorie zbiegają do $x_0$;
* punkt stały $x_0$ jest **repelerem**, gdy punkty startujące dostatecznie blisko $x_0$, ale różne od niego, są od niego odpychane.

---

### Najpierw opiszmy iteracje

Weźmy dowolne $x_0\in[0,1]$.

#### Gdy $x_0$ jest niewymierne

Wtedy
$$
f(x_0)=x_0,
$$
więc trajektoria jest stała:
$$
x_n=x_0 \quad \text{dla każdego } n.
$$

#### Gdy $x_0$ jest wymierne

Wtedy
$$
x_1=f(x_0)=\frac{x_0}{\pi}.
$$

Ponieważ:

* $x_0\in\mathbb{Q}$,
* $\pi$ jest niewymierne,
* dla $x_0\neq 0$ iloraz $\frac{x_0}{\pi}$ jest niewymierny,

to dla każdego wymiernego $x_0>0$ mamy
$$
x_1=\frac{x_0}{\pi}\in\mathbb{R}\setminus\mathbb{Q}.
$$

A zatem już od następnego kroku:
$$
x_2=f(x_1)=x_1,
$$
czyli trajektoria staje się stała.

Podsumowując:

* jeśli $x_0=0$, to $x_n=0$,
* jeśli $x_0\in\mathbb{Q}\cap(0,1]$, to
  $$
  x_n=\frac{x_0}{\pi} \quad \text{dla każdego } n\ge 1,
  $$
  i granicą jest $\frac{x_0}{\pi}$,
* jeśli $x_0$ jest niewymierne, to $x_n=x_0$ dla każdego $n$.

---

### Czy $0$ jest atraktorem?

Nie.

Rzeczywiście, w każdym otoczeniu punktu $0$ są liczby niewymierne $y>0$, a dla nich
$$
f(y)=y,
$$
więc trajektoria startująca z $y$ pozostaje równa $y$ i nie zbiega do $0$.

Zatem nie istnieje otoczenie punktu $0$, z którego wszystkie trajektorie zmierzałyby do $0$.

Czy $0$ jest repelerem? Też nie.

Bo w każdym jego otoczeniu są punkty niewymierne bardzo bliskie $0$, a ich trajektorie w ogóle się od $0$ nie oddalają — po prostu stoją w miejscu. Dodatkowo dla wymiernych $q>0$ bliskich $0$ mamy
$$
f(q)=\frac{q}{\pi},
$$
co nadal jest blisko $0$, a nie „ucieka” z małego otoczenia.

Zatem:
$$
0 \text{ nie jest ani atraktorem, ani repelerem.}
$$

---

### Czy niewymierny punkt stały $x_0\in[0,1]\setminus\mathbb{Q}$ jest atraktorem?

Nie.

W każdym otoczeniu $x_0$ są inne punkty niewymierne $y\neq x_0$, a dla nich trajektoria jest stała:
$$
y_n=y.
$$

Taka trajektoria nie zbiega do $x_0$.

Są też punkty wymierne $q$ bliskie $x_0$, ale ich trajektorie zbiegają do
$$
\frac{q}{\pi},
$$
co na ogół nie jest równe $x_0$.

Więc żaden niewymierny punkt stały nie jest atraktorem.

---

### Czy niewymierny punkt stały $x_0$ jest repelerem?

Też nie.

Żeby punkt był repelerem, punkty startujące blisko niego powinny być z niego „wypychane”. Tymczasem w każdym otoczeniu $x_0$ są niewymierne punkty $y\neq x_0$, dla których
$$
y_n=y
$$
dla każdego $n$, więc pozostają one dowolnie blisko $x_0$.

Zatem taki punkt nie jest repelerem.

---

### Wniosek do punktu b)

Żaden punkt stały nie jest atraktorem ani repelerem.

Czyli:
$$
\text{dla każdego } x_0\in Fix(f), \quad x_0 \text{ nie jest ani atraktorem, ani repelerem.}
$$

---

## c) Wyznaczenie zbiorów $Bas_f(x)$ dla każdego $x\in[0,1]$

Musimy określić, dla jakich punktów początkowych $x_0$ trajektoria zbiega do danego $x$.

Z poprzedniej analizy wiemy:

1. jeśli $x_0$ jest niewymierne, to
   $$
   x_n=x_0 \quad \text{dla każdego } n,
   $$
   więc trajektoria zbiega do $x_0$;

2. jeśli $x_0=0$, to
   $$
   x_n=0,
   $$
   więc trajektoria zbiega do $0$;

3. jeśli $x_0\in\mathbb{Q}\cap(0,1]$, to
   $$
   x_n=\frac{x_0}{\pi} \quad \text{dla każdego } n\ge 1,
   $$
   więc trajektoria zbiega do $\frac{x_0}{\pi}$.

---

### Basen punktu $0$

Do $0$ zbiega tylko trajektoria startująca z $0$.

Bo:

* niewymierne dodatnie pozostają sobą,
* wymierne dodatnie przechodzą do $\frac{x_0}{\pi}>0$.

Zatem
$$
Bas_f(0)={0}.
$$

---

### Basen punktu niewymiernego $x\in[0,1]\setminus\mathbb{Q}$

Szukamy wszystkich $x_0$, dla których trajektoria zbiega do $x$.

Na pewno należy sam punkt $x$, bo jeśli startujemy z $x$, to trajektoria jest stała.

Czy mogą być jeszcze jakieś inne punkty?

* jeśli $x_0$ jest niewymierne i $x_0\neq x$, to trajektoria zbiega do $x_0$, nie do $x$;
* jeśli $x_0$ jest wymierne dodatnie, to trajektoria zbiega do $\frac{x_0}{\pi}$.

Aby granica była równa $x$, musimy mieć
$$
\frac{x_0}{\pi}=x,
$$
czyli
$$
x_0=\pi x.
$$

Taki punkt może wejść do basenu tylko wtedy, gdy:

* $\pi x\in[0,1]$,
* $\pi x$ jest wymierne.

Drugi warunek jest nietrywialny, ale możliwy dla pewnych niewymiernych $x$, np. gdy $x=\frac{r}{\pi}$ z wymiernym $r\in(0,1]$.

Zatem dla niewymiernego $x$ mamy:
$$
Bas_f(x)=
\begin{cases}
{x,\pi x}, & \text{gdy } \pi x\in\mathbb{Q}\cap(0,1],[1ex]
{x}, & \text{w przeciwnym razie.}
\end{cases}
$$

Można to zapisać krócej jako
$$
Bas_f(x)={x}\cup\Bigl({\pi x}\cap\mathbb{Q}\cap[0,1]\Bigr),
\qquad x\in[0,1]\setminus\mathbb{Q}.
$$

---

### Basen punktu wymiernego $x\in(0,1]\cap\mathbb{Q}$

Taki punkt **nie jest punktem stałym**, a z analizy trajektorii wynika, że żadna trajektoria nie może do niego zbiegać:

* trajektorie startujące z niewymiernych są stałe i niewymierne,
* trajektorie startujące z wymiernych dodatnich po jednym kroku przechodzą do punktu niewymiernego,
* jedynie $0$ daje granicę $0$.

Zatem dla dodatniego wymiernego $x$:
$$
Bas_f(x)=\emptyset.
$$

---

## Ostateczny wynik

### Punkty stałe

$$
Fix(f)=\left([0,1]\setminus\mathbb{Q}\right)\cup{0}.
$$

### Atraktory i repelery

Żaden punkt stały nie jest ani atraktorem, ani repelerem.

### Baseny

* dla $x=0$:
  $$
  Bas_f(0)={0};
  $$

* dla $x\in[0,1]\setminus\mathbb{Q}$:
  $$
  Bas_f(x)={x}\cup\Bigl({\pi x}\cap\mathbb{Q}\cap[0,1]\Bigr);
  $$

* dla $x\in(0,1]\cap\mathbb{Q}$:
  $$
  Bas_f(x)=\emptyset.
  $$

## Zadanie 5

Rozważamy układ dynamiczny

x_{n+1}=f(x_n), \qquad f:\mathbb{R}\to\mathbb{R}, \qquad f(x)=\sqrt{x^2+x+1}.
$$

Mamy:

* wyznaczyć punkty stałe,
* pokazać, że $f(x)>x$ dla każdego $x\in\mathbb{R}$,
* udowodnić, że każda trajektoria jest rosnąca i nieograniczona,
* wykazać, że $+\infty$ jest atraktorem układu.

---

## a) Wyznaczenie zbioru punktów stałych $Fix(f)$

Szukamy rozwiązań równania
$$
f(x)=x,
$$
czyli
$$
\sqrt{x^2+x+1}=x.
$$

Lewa strona jest nieujemna, więc koniecznie musi być
$$
x\ge 0.
$$

Podnosząc równanie do kwadratu, dostajemy
$$
x^2+x+1=x^2.
$$

Stąd
$$
x+1=0,
$$
czyli
$$
x=-1.
$$

Ale to przeczy warunkowi $x\ge 0$. Zatem równanie nie ma rozwiązań rzeczywistych.

Wniosek:
$$
Fix(f)=\emptyset.
$$

---

## b) Dowód, że dla każdego $x\in\mathbb{R}$ zachodzi $f(x)>x$

Chcemy pokazać, że
$$
\sqrt{x^2+x+1}>x
\qquad \text{dla każdego } x\in\mathbb{R}.
$$

Trzeba rozpatrzyć dwa przypadki.

### Przypadek 1: $x<0$

Wtedy
$$
f(x)=\sqrt{x^2+x+1}\ge 0>x.
$$

Zatem nierówność jest prawdziwa.

### Przypadek 2: $x\ge 0$

Wtedy obie strony są nieujemne, więc można bezpiecznie podnieść nierówność do kwadratu:
$$
\sqrt{x^2+x+1}>x
\iff x^2+x+1>x^2.
$$

To daje
$$
x+1>0,
$$
co jest prawdą dla każdego $x\ge 0$.

Zatem rzeczywiście
$$
f(x)>x \qquad \text{dla każdego } x\in\mathbb{R}.
$$

---

## c) Dowód, że każda trajektoria $(f^n(x))_{n\ge 0}$ jest rosnąca i nieograniczona

Niech
$$
x_{n+1}=f(x_n)=\sqrt{x_n^2+x_n+1}.
$$

Z punktu b) wiemy, że dla każdego argumentu $t\in\mathbb{R}$ zachodzi
$$
f(t)>t.
$$

Podstawiając $t=x_n$, dostajemy
$$
x_{n+1}=f(x_n)>x_n
\qquad \text{dla każdego } n\ge 0.
$$

To znaczy, że ciąg $(x_n)$ jest **ściśle rosnący**.

Pozostaje wykazać, że jest **nieograniczony z góry**.

Załóżmy przeciwnie, że $(x_n)$ jest ograniczony z góry. Wtedy jako ciąg rosnący byłby zbieżny. Niech
$$
x_n\to g.
$$

Przechodząc do granicy w rekurencji
$$
x_{n+1}=\sqrt{x_n^2+x_n+1},
$$
otrzymujemy
$$
g=\sqrt{g^2+g+1}.
$$

To oznacza, że $g$ byłby punktem stałym funkcji $f$. Ale w punkcie a) pokazaliśmy, że funkcja nie ma punktów stałych.

Sprzeczność.

Zatem ciąg $(x_n)$ nie może być ograniczony z góry. Wobec tego każda trajektoria jest:

* ściśle rosnąca,
* nieograniczona z góry.

Czyli dla każdego $x_0\in\mathbb{R}$:
$$
x_n\to +\infty.
$$

---

## d) Wykazanie, że $+\infty$ jest atraktorem układu

Z punktu c) wynika, że dla każdego punktu początkowego $x_0\in\mathbb{R}$ trajektoria spełnia
$$
x_n\to +\infty.
$$

To znaczy, że wszystkie trajektorie są przyciągane do $+\infty$.

Zatem w sensie dynamiki na prostej rozszerzonej
$$
\overline{\mathbb{R}}=\mathbb{R}\cup{+\infty,-\infty},
$$
punkt
$$
+\infty
$$
jest **globalnym atraktorem** układu.

Można więc napisać:
$$
Bas_f(+\infty)=\mathbb{R}.
$$

## Zadanie 6

Rozważamy układ dynamiczny
$$
x_{n+1}=f(x_n), \qquad f:[0,1]\to[0,1], \qquad f(x)=1-|2x-1|.
$$

Równoważnie:
$$
f(x)=
\begin{cases}
2x, & x\in\left[0,\frac12\right],[1ex]
2-2x, & x\in\left[\frac12,1\right].
\end{cases}
$$

## a) Punkty stałe $Fix(f)$ i ich stabilność

Szukamy rozwiązań równania
$$
f(x)=x.
$$

Rozpatrujemy dwa przypadki.

### Przypadek 1: $x\in\left[0,\frac12\right]$

Wtedy
$$
f(x)=2x.
$$

Warunek punktu stałego daje
$$
2x=x,
$$
czyli
$$
x=0.
$$

### Przypadek 2: $x\in\left[\frac12,1\right]$

Wtedy
$$
f(x)=2-2x.
$$

Warunek punktu stałego daje
$$
2-2x=x,
$$
czyli
$$
2=3x,
$$
a więc
$$
x=\frac23.
$$

Zatem
$$
Fix(f)=\{0,\frac23\}.
$$

### Stabilność

Dla $x\neq \frac12$ mamy
$$
f'(x)=
\begin{cases}
2, & x<\frac12,[1ex]
-2, & x>\frac12.
\end{cases}
$$

W szczególności:

* w punkcie $0$:
  $$
  |f'(0)|=2>1,
  $$
* w punkcie $\frac23$:
  $$
  \left|f'\left(\frac23\right)\right|=|-2|=2>1.
  $$

Oba punkty stałe są więc **repelerami**.

Wniosek:
$$
Fix(f)=\{0,\frac23\},
\qquad
0 \text{ i } \frac23 \text{ są repelerami.}
$$

## b) Wszystkie punkty 2-okresowe

Szukamy punktów spełniających
$$
f^2(x)=x,
$$
które **nie są** punktami stałymi funkcji $f$.

Najpierw policzmy $f^2$ kawałkami.

Ponieważ
$$
f(x)=
\begin{cases}
2x, & x\in\left[0,\frac12\right],\
2-2x, & x\in\left[\frac12,1\right],
\end{cases}
$$
trzeba rozbić przedział na mniejsze części.

---

### Dla $x\in\left[0,\frac14\right]$

Wtedy
$$
f(x)=2x\in\left[0,\frac12\right],
$$
więc
$$
f^2(x)=2(2x)=4x.
$$

Warunek $f^2(x)=x$ daje
$$
4x=x \iff 3x=0 \iff x=0.
$$

To punkt stały, więc go później odrzucimy.

---

### Dla $x\in\left[\frac14,\frac12\right]$

Wtedy
$$
f(x)=2x\in\left[\frac12,1\right],
$$
więc
$$
f^2(x)=2-2(2x)=2-4x.
$$

Warunek $f^2(x)=x$ daje
$$
2-4x=x \iff 2=5x \iff x=\frac25.
$$

To nie jest punkt stały $f$, więc jest kandydatem 2-okresowym.

---

### Dla $x\in\left[\frac12,\frac34\right]$

Wtedy
$$
f(x)=2-2x\in\left[\frac12,1\right],
$$
więc
$$
f^2(x)=2-2(2-2x)=4x-2.
$$

Warunek $f^2(x)=x$ daje
$$
4x-2=x \iff 3x=2 \iff x=\frac23.
$$

To punkt stały, więc nie jest punktem 2-okresowym właściwym.

---

### Dla $x\in\left[\frac34,1\right]$

Wtedy
$$
f(x)=2-2x\in\left[0,\frac12\right],
$$
więc
$$
f^2(x)=2(2-2x)=4-4x.
$$

Warunek $f^2(x)=x$ daje
$$
4-4x=x \iff 4=5x \iff x=\frac45.
$$

To nie jest punkt stały, więc jest drugim punktem 2-okresowym.

---

Zatem rozwiązania równania $f^2(x)=x$ to
$$
0,\quad \frac25,\quad \frac23,\quad \frac45.
$$

Po odrzuceniu punktów stałych $0$ i $\frac23$ zostają punkty 2-okresowe:
$$
\frac25,\quad \frac45.
$$

Rzeczywiście:

f\left(\frac25\right)=\frac45,\qquad
f\left(\frac45\right)=\frac25.

$$
\{\frac25,\frac45\}

jest orbitą 2-okresową.

## c) Dla każdego $k\in\mathbb{N}$ funkcja $f^k$ ma dokładnie $2^k$ punktów stałych

To jest standardowa własność mapy namiotowej o nachyleniach $\pm 2$.

### Krok 1: struktura funkcji $f^k$

Każda iteracja dzieli przedział $[0,1]$ na coraz drobniejsze kawałki. Dokładniej, funkcja $f^k$ jest liniowa na każdym z przedziałów
$$
\left[\frac{j}{2^k},\frac{j+1}{2^k}\right], \qquad j=0,1,\dots,2^k-1,
$$
i na każdym takim przedziale ma nachylenie
$$
\pm 2^k.
$$

Czyli wykres $f^k$ składa się z dokładnie $2^k$ odcinków liniowych.

### Krok 2: przecięcia z prostą $y=x$

Na każdym takim odcinku funkcja
$$
f^k(x)-x
$$
jest liniowa o nachyleniu
$$
\pm 2^k -1,
$$
czyli nigdy nie ma nachylenia zerowego. Zatem na każdym odcinku może być **co najwyżej jedno** przecięcie z prostą $y=x$.

Teraz istotne jest, że każdy z tych $2^k$ odcinków przechodzi od jednej krawędzi kwadratu jednostkowego do drugiej w taki sposób, że prostą $y=x$ przecina dokładnie raz. Intuicyjnie: każdy „ząb” wykresu $f^k$ przecina przekątną raz.

Wobec tego liczba punktów stałych funkcji $f^k$ wynosi dokładnie
$$
2^k.
$$

Czyli:
$$
Fix( f^k ) = 2^k.
$$

### Wniosek o orbitach dowolnie wysokiego okresu

Skoro dla każdego $k$ funkcja $f^k$ ma punkty stałe, to istnieją punkty okresowe okresu dzielącego $k$.

Ale tu można powiedzieć więcej: dla mapy namiotowej pojawiają się punkty o dowolnym okresie naturalnym. Wynika to stąd, że liczba punktów stałych iteracji rośnie wykładniczo, więc nie mogą one wszystkie pochodzić wyłącznie z okresów mniejszych. Zatem dla każdego
$$
n\in\mathbb{N}
$$
istnieje orbita o okresie dokładnie $n$.

Wniosek:
$$
\text{układ ma orbity o dowolnie dużym okresie naturalnym.}
$$

## d) Czułość na warunki początkowe

Mamy wykazać, że dla dowolnych $x,y\in[0,1]$, $x\neq y$, istnieje $n\in\mathbb{N}$ takie, że
$$
|f^n(x)-f^n(y)|\ge \frac12.
$$

To jest istota chaosu dla mapy namiotowej.

### Idea

Funkcja $f$ na każdym kawałku rozciąga odległości dwukrotnie:

* na lewym ramieniu ma nachylenie $2$,
* na prawym ramieniu ma nachylenie $-2$.

Dopóki dwa punkty $x,y$ pozostają w tym samym przedziale monotoniczności iteracji $f^n$, ich odległość po $n$ krokach rośnie jak
$$
|f^n(x)-f^n(y)|=2^n|x-y|.
$$

Nie mogą więc pozostawać blisko siebie w nieskończoność, bo długość przedziału jest skończona. W pewnym momencie obrazy tych punktów muszą znaleźć się po przeciwnych stronach wierzchołka albo po prostu zostać rozseparowane na odległość co najmniej $\frac12$.

Formalniej: przedziały monotoniczności dla $f^n$ mają długość
$$
\frac{1}{2^n}.
$$
Jeśli wybierzemy $n$ tak, by
$$
\frac{1}{2^n}<|x-y|,
$$
to punkty $x$ i $y$ nie mogą należeć do tego samego przedziału monotoniczności funkcji $f^n$.

A końce tych przedziałów są odwzorowywane przez $f^n$ naprzemiennie na $0$ i $1$, więc obrazy dwóch różnych sąsiednich gałęzi są mocno rozdzielone. W szczególności dostajemy
$$
|f^n(x)-f^n(y)|\ge \frac12.
$$

To kończy dowód czułości.

Wniosek:
$$
\text{układ jest czuły na warunki początkowe.}
$$

* punkty stałe są repelerami,
* układ jest czuły na warunki początkowe,
* trajektorie nie skupiają się wokół jednego punktu.

Gdyby istniał punktowy atraktor $a\in[0,1]$, przyciągający prawie wszystkie trajektorie, to większość trajektorii musiałaby spełniać
$$
f^n(x)\to a.
$$
To jest sprzeczne z istnieniem gęstego i bardzo bogatego zbioru punktów okresowych oraz z rozciągającym charakterem układu.

Zatem:
$$
\text{w tym układzie nie istnieje punktowy atraktor przyciągający prawie wszystkie trajektorie.}
$$

---

## Ostateczne podsumowanie

### Punkty stałe

$$
Fix(f)=\{0,\frac23\}.
$$

### Stabilność

Oba punkty stałe są repelerami, bo
$$
|f'(0)|=2,\qquad \left|f'\left(\frac23\right)\right|=2.
$$

### Punkty 2-okresowe

$$
\frac25,\qquad \frac45.
$$

Tworzą orbitę:
$$
\frac25 \mapsto \frac45 \mapsto \frac25.
$$

### Punkty stałe iteracji

Dla każdego $k\in\mathbb{N}$ funkcja $f^k$ ma dokładnie
$$
2^k
$$
punktów stałych.

Stąd wynika istnienie orbit o dowolnym okresie naturalnym.

### Czułość

Dla dowolnych $x\neq y$ istnieje $n$ takie, że
$$
|f^n(x)-f^n(y)|\ge \frac12.
$$

### Atraktor punktowy

Nie istnieje punktowy atraktor przyciągający prawie wszystkie trajektorie.
