# 🔷 Metoda Neldera–Meada – intuicyjne wyjaśnienie

Metoda **Neldera–Meada** to jedna z najprostszych i najbardziej intuicyjnych metod **optymalizacji bez użycia pochodnych**.  
Oznacza to, że potrafi znaleźć minimum funkcji **nawet wtedy, gdy nie umiemy obliczyć jej gradientu** (pochodnych cząstkowych).  
Dlatego często stosuje się ją w praktyce – w fizyce, uczeniu maszynowym czy dopasowywaniu modeli.

---

## 🧩 Pomysł ogólny

Zamiast poruszać się wzdłuż kierunku gradientu (jak w metodach gradientowych), Nelder–Mead używa **grupy punktów** zwanej **simpleksem**.

- Dla funkcji dwóch zmiennych (x, y) simpleks to **trójkąt**.  
- Dla trzech zmiennych – **czworościan**,  
- a ogólnie: dla $ n $ zmiennych to figura o $ n + 1 $ wierzchołkach.

Każdy wierzchołek to **punkt w przestrzeni rozwiązań**, w którym obliczamy wartość funkcji celu $ f(x) $.

---

## 🔁 Jak działa metoda krok po kroku

1. **Inicjalizacja simpleksu**  
   - Wybieramy punkt startowy $ x_0 $ i tworzymy wokół niego pozostałe $ n $ wierzchołków (np. lekko przesuniętych).  
   - Otrzymujemy $ n+1 $ punktów: $ x_1, x_2, \dots, x_{n+1} $.  
   - Dla każdego obliczamy wartość funkcji $ f(x_i) $.

2. **Sortowanie punktów**  
   - Porządkujemy punkty od najlepszego (najmniejsza wartość $ f $) do najgorszego (największa wartość $ f $).  
   - Oznaczamy:
     - $ x_{\text{best}} $ – najlepszy punkt,
     - $ x_{\text{worst}} $ – najgorszy punkt,
     - $ x_{\text{centroid}} $ – środek pozostałych punktów (bez najgorszego).

3. **Odbicie (reflection)**  
   - Chcemy sprawdzić, czy po drugiej stronie środka nie ma lepszego miejsca.  
   - Tworzymy nowy punkt:
     $$
     x_r = x_{\text{centroid}} + \alpha (x_{\text{centroid}} - x_{\text{worst}})
     $$
     gdzie $ \alpha > 0 $ to współczynnik odbicia (zwykle 1).  
   - Jeśli $ f(x_r) $ jest lepsze niż większość punktów, to zastępujemy $ x_{\text{worst}} $ nowym punktem.

4. **Ekspansja (expansion)**  
   - Jeśli odbicie dało **dużo lepszy wynik** (czyli nowy punkt jest nawet lepszy niż dotychczasowy najlepszy),  
     to próbujemy iść jeszcze dalej w tym kierunku:
     $$
     x_e = x_{\text{centroid}} + \gamma (x_r - x_{\text{centroid}})
     $$
     gdzie $ \gamma > 1 $ (np. 2).  
   - Jeśli ekspansja przynosi poprawę, zostawiamy $ x_e $; w przeciwnym razie pozostajemy przy $ x_r $.

5. **Kontrakcja (contraction)**  
   - Jeśli odbicie **nie pomogło** (czyli nowy punkt jest gorszy niż dotychczasowe),  
     cofamy się w stronę środka:
     $$
     x_c = x_{\text{centroid}} + \rho (x_{\text{worst}} - x_{\text{centroid}})
     $$
     gdzie $ 0 < \rho < 1 $ (np. 0.5).  
   - Jeśli ten punkt jest lepszy niż najgorszy, przyjmujemy go;  
     w przeciwnym razie przechodzimy do następnego kroku.

6. **Redukcja (shrink)**  
   - Jeśli nic nie zadziałało, zmniejszamy cały simpleks wokół najlepszego punktu:
     $$
     x_i = x_{\text{best}} + \sigma (x_i - x_{\text{best}}), \quad \text{dla } i = 2, \dots, n+1
     $$
     gdzie $ \sigma \approx 0.5 $.  
   - Dzięki temu wszystkie punkty zbliżają się do najlepszego, a obszar poszukiwań się „kurczy”.

7. **Kryterium stopu**  
   - Algorytm zatrzymuje się, gdy:
     - wszystkie punkty simpleksu są bardzo blisko siebie,  
     - lub różnice wartości funkcji $ f(x_i) $ są bardzo małe,  
     - albo osiągnięto maksymalną liczbę iteracji.

---

## ⚙️ Co robi metoda w praktyce

Metoda Neldera–Meada **porusza się i deformuje simpleks**:  
czasem go **odbija**, czasem **rozciąga**, a czasem **kurczy**, szukając coraz lepszego obszaru.  
Dzięki temu **nie wymaga żadnych pochodnych** – wystarczy umieć obliczyć wartość funkcji $ f(x) $.

---

## 📉 Zalety

✅ Działa nawet dla funkcji:
- niegładkich (bez pochodnych),
- zaszumionych lub trudnych do opisania analitycznie,  
- z nieznanymi ograniczeniami.

✅ Łatwa do zaimplementowania.  
✅ Działa w dowolnej liczbie wymiarów.

---

## ⚠️ Wady

❌ Może **utknąć w minimum lokalnym** (nie zawsze znajduje globalne).  
❌ Bywa **wolna**, szczególnie przy dużej liczbie zmiennych.  
❌ Nie gwarantuje zbieżności dla wszystkich typów funkcji.

---

## 🧠 Podsumowanie

Metoda Neldera–Meada to **inteligentne testowanie wartości funkcji** w kilku punktach jednocześnie.  
Zamiast „zjeżdżać w dół” po gradientach, buduje ona i przesuwa **wielokąt (simpleks)** po krajobrazie funkcji,  
aż znajdzie miejsce, w którym wartości funkcji przestają się znacząco poprawiać.

W skrócie:

| Etap | Co się dzieje | Współczynnik |
|------|----------------|---------------|
| Odbicie | Przesunięcie w stronę lepszego obszaru | α ≈ 1 |
| Ekspansja | Idź dalej, jeśli lepiej | γ ≈ 2 |
| Kontrakcja | Cofnij się, jeśli gorzej | ρ ≈ 0.5 |
| Redukcja | Zmniejsz cały simpleks | σ ≈ 0.5 |

---

📌 **Podsumowując:**  
Metoda Neldera–Meada to prosty, intuicyjny i bardzo uniwersalny sposób szukania minimum funkcji.  
Nie potrzebuje pochodnych, więc nadaje się do problemów, w których jedyne, co umiemy zrobić, to **zmierzyć wartość funkcji w danym punkcie**.


```python
def rosenbrock(x):
    """f(x1,x2) = (1-x1)^2 + 100*(x2 - x1^2)^2"""
    x = np.asarray(x, dtype=float)
    return (1 - x[0])**2 + 100.0 * (x[1] - x[0]**2)**2

def himmelblau(x):
    """Funkcja Himmelblaua."""
    x, y = x
    return (x**2 + y - 11)**2 + (x + y**2 - 7)**2
```

```python
def nelder_mead(f, x0, alpha=1.0, gamma=2.0, rho=0.5, sigma=0.5,
                tol=1e-6, max_iter=500):
    """
    Bardzo prosta implementacja metody Neldera–Meada (bez pochodnych).
    Zwraca (x_best, f_best, trajektoria, simplex_hist).
    """
    x0 = np.asarray(x0, dtype=float)
    n = x0.size

    simplex = [x0]
    step = 0.05  
    for i in range(n):
        x = x0.copy()
        x[i] = x[i] + step if x[i] != 0 else step
        simplex.append(x)
    simplex = np.array(simplex)

    path = [simplex[0].copy()]
    simplex_hist = [simplex.copy()]  # zapamiętuj historię simpleksu

    for _ in range(max_iter):
        # sortuj wierzchołki po wartości f
        simplex = np.array(sorted(simplex, key=f))
        f_vals = np.array([f(x) for x in simplex])

        # kryterium stopu: rozrzut wartości f
        if f_vals.std() < tol:
            break

        x_best = simplex[0]
        x_worst = simplex[-1]
        centroid = simplex[:-1].mean(axis=0)

        # odbicie
        x_r = centroid + alpha * (centroid - x_worst)
        f_r = f(x_r)

        if f_r < f_vals[0]:
            # ekspansja
            x_e = centroid + gamma * (x_r - centroid)
            simplex[-1] = x_e if f(x_e) < f_r else x_r
        elif f_vals[0] <= f_r < f_vals[-2]:
            # akceptuj odbicie
            simplex[-1] = x_r
        else:
            # kontrakcja
            x_c = centroid + rho * (x_worst - centroid)
            if f(x_c) < f_vals[-1]:
                simplex[-1] = x_c
            else:
                # redukcja do najlepszego
                for i in range(1, n+1):
                    simplex[i] = x_best + sigma * (simplex[i] - x_best)

        path.append(simplex[0].copy())
        simplex_hist.append(simplex.copy())

    x_best = simplex[0]
    return x_best, f(x_best), np.array(path), np.array(simplex_hist)    
```

```python
import scipy.optimize as opt
import matplotlib.pyplot as plt
```

```python
res = opt.minimize(himmelblau, x0=[0.0, 0.0], method="Nelder-Mead",
               options={"disp": True, "maxiter": 500})
print(res.x, '\n', res.fun, '\n', res.nit, '\n', res.nfev, '\n', res.success)
```

```python
x0 = np.array([0.0, 0.0])   # punkt startowy
x_min, f_min, traj = nelder_mead(himmelblau, x0)
print("Minimum w punkcie:", x_min)
print("Wartość funkcji:", f_min)
```

```python
plt.plot(traj[:, 0], traj[:, 1], marker='o')
```

# Metoda Gradientowa

```python
def rosenbrock(x):
    """Funkcja Rosenbrocka f(x,y) = (1-x)^2 + 100*(y-x^2)^2"""
    x1, x2 = x
    return (1 - x1)**2 + 100 * (x2 - x1**2)**2

def himmelblau(x):
    """Funkcja Himmelblaua."""
    x, y = x
    return (x**2 + y - 11)**2 + (x + y**2 - 7)**2

def rosenbrock_grad(x):
    """Gradient funkcji Rosenbrocka"""
    x1, x2 = x
    df_dx1 = -2*(1 - x1) - 400*x1*(x2 - x1**2)
    df_dx2 = 200*(x2 - x1**2)
    return np.array([df_dx1, df_dx2])

def himmelblau_grad(x):
    """Gradient funkcji Himmelblaua"""
    x, y = x
    df_dx = 4*x*(x**2 + y - 11) + 2*(x + y**2 - 7)
    df_dy = 2*(x**2 + y - 11) + 4*y*(x + y**2 - 7)
    return np.array([df_dx, df_dy])

def gradient_descent(f, grad_f, x0, lr=0.001, tol=1e-6, max_iter=10000):
    """
    Prosta implementacja metody spadku gradientowego (gradient descent)
    f       - funkcja celu
    grad_f  - gradient funkcji
    x0      - punkt startowy (np.array)
    lr      - learning rate (krok uczenia)
    tol     - tolerancja błędu
    max_iter - maks. liczba iteracji
    """
    x = np.array(x0, dtype=float)
    path = [x.copy()]  # historia punktów

    for i in range(max_iter):
        grad = grad_f(x)
        x_new = x - lr * grad
        path.append(x_new.copy())

        # warunek stopu
        if np.linalg.norm(x_new - x) < tol:
            break
        x = x_new

    return x, f(x), np.array(path)
```

```python
x0 = np.array([-1.5, 2.0])  # start z górnej części doliny
x_min, f_min, traj = gradient_descent(rosenbrock, rosenbrock_grad, x0,
                                          lr=0.001, max_iter=20000)
print("Minimum znalezione w:", x_min)
print("Wartość funkcji:", f_min)
plt.plot(traj[:, 0], traj[:, 1], marker='o')
plt.title("Trajektoria spadku gradientowego")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
x0 = np.array([0.0, 0.0])  # start blisko jednego z minimów
x_min, f_min, traj = gradient_descent(himmelblau, himmelblau_grad, x0,
                                          lr=0.001, max_iter=20000)
print("Minimum znalezione w:", x_min)
print("Wartość funkcji:", f_min)
plt.plot(traj[:, 0], traj[:, 1], marker='o')
plt.title("Trajektoria spadku gradientowego")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
# Animacja Neldera–Meada dla funkcji Himmelblaua
from matplotlib import animation
from IPython.display import HTML
import numpy as np
# Uruchom algorytm z ograniczoną liczbą iteracji (dla płynnej animacji)
x0 = np.array([0.0, 0.0])
x_min, f_min, path, simplex_hist = nelder_mead(himmelblau, x0, max_iter=200)

# Siatka do konturów
xs = np.linspace(-6, 6, 300)
ys = np.linspace(-6, 6, 300)
X, Y = np.meshgrid(xs, ys)
Z = np.vectorize(lambda a, b: himmelblau([a, b]))(X, Y)

fig, ax = plt.subplots(figsize=(6,6))
cs = ax.contour(X, Y, Z, levels=40, cmap='viridis')
ax.set_xlim(xs.min(), xs.max())
ax.set_ylim(ys.min(), ys.max())
ax.set_title('Nelder–Mead: animacja simpleksu na tle konturów (Himmelblau)')

# Linie reprezentujące krawędzie simpleksu (dla 2D mamy 3 krawędzie)
simplex_lines = [ax.plot([], [], 'r-', lw=2)[0] for _ in range(3)]
# Trajektoria najlepszego punktu
traj_line, = ax.plot([], [], 'bo-', markersize=4)

def init():
    traj_line.set_data([], [])
    for ln in simplex_lines:
        ln.set_data([], [])
    return [traj_line] + simplex_lines

def update(i):
    s = simplex_hist[min(i, len(simplex_hist)-1)]  # (n+1, 2)
    # n+1 punktów simpleksu, połączymy je w pętlę
    for k, ln in enumerate(simplex_lines):
        a = s[k]
        b = s[(k+1) % s.shape[0]]
        ln.set_data([a[0], b[0]], [a[1], b[1]])
    traj_line.set_data(path[:i+1, 0], path[:i+1, 1])
    return [traj_line] + simplex_lines

anim = animation.FuncAnimation(fig, update, frames=len(simplex_hist), init_func=init,
                               blit=True, interval=200)

# Wyświetl animację inline
HTML(anim.to_jshtml())
```
